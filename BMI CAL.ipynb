```python
name=input('Enter your name')
weight = float(input("Input your weight in pounds: "))
height = float(input("Input your height in inches: "))

bmi = (weight / (height ** 2)) * 703

print("Your BMI is:", bmi)
```

```python
if bmi>0:
    if(bmi<18.5):
        print(name,'You are under weight')
    elif((bmi>=18.6) & (bmi<=24.9)):
        print(name,'You are normal Weighted')
    elif((bmi>=25.0) & (bmi<=29.9)):
        print(name,'You are over Weighted')
    else:
        print('You are obese')
else:
    print('Enter a valid Number')
```
