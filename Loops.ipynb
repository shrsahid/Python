For Loop

```python
ice_cream_dict ={'name':'Alex Freberg','weekly':5,'favorite ice creams':['MCC','Chocolate']}


for key,value in ice_cream_dict.items():
    print(key,'->',value)
```

```python
ice_cream_dict ={'name':'Alex Freberg','weekly':5,'favorite ice creams':['MCC','Chocolate']}


for cream in ice_cream_dict.values():
    print(cream)
```

Nested For Loop

```python
flavors=['vanila','choclate']
toppings=['Hot Fudge','Oreos']


for one in flavors:
    for two in toppings:
        print(one,'topped with',two)
```

While Loop

```python
num=1

while num<5:
    print(num)
    num = num +1
```

```python
num=1

while num<5:
    print(num)
    if num == 6:
        break
    num = num +1
else:
    print('No longer 5')
```

```python
while num<5:
    num = num +1
    if num == 3:
        continue
    print(num)
else:
    print('No longer 5')
```
